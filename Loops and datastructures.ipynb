```python
# 3 names in a list and find the vowels:

v = {'a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U'}
l = ["Dibyajit", "Arijit", "Sujit"]

l2 = set("".join(l))

l3 = [x for x in v.intersection(l2)]
print (l3)
```
