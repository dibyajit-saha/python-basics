```python
### Area of a circle using math module
import math

r = int (input ("Enter the radius of the circle in cm: "))
area = math.pi * r * r
print (f"The area of the circle is {area} cm^2")
```

```python
## ATM machine

balance = 1000.0

def check_balance():
    print (f'Your current balance is {balance}')

def deposit():
    global balance
    amount = float(input ("Enter the amount to be deposited: "))
    if amount > 0:
        balance = balance + amount
        print (f'{amount} has been deposited.')
    else: 
        print ("Invalid amount")

def withdraw():
    global balance
    amount = float(input ("Enter the amount to be withdrawn: "))
    if 0 < amount <= balance:
        balance = balance - amount
        print (f'{amount} has been withdrawn.')
    else: 
        print ("Invalid amount or Insufficient Balance!")

def atm ():
    print ("----------Welcome to the ATM--------------")
    while True:
        print ("\n --------Menu---------")
        print ("1. Check Balance")
        print ("2. Deposit")
        print ("3. Withdraw")
        print ("4. Exit")

        choice = int (input ("Enter your choice (1-4): "))
        if choice == 1:
            check_balance()
        elif choice == 2:
            deposit()
        elif choice == 3:
            withdraw()
        elif choice == 4:
            print ("Thank you for using the ATM. Good Bye!!!")
            break
        else:
            print ("Invalid choice. Please Choose from 1-4")
```

```python
atm()
```
