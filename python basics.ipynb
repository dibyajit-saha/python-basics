```python
n1 = int(input("Entre the first number: "))
n2 = int(input("Entre the second number: "))
n3 = int(input("Entre the third number: "))

if n1>n2 and n1>n3:
    print (f" The highest number is {n1}")
elif n2>n1 and n2>n3:
    print (f" The highest number is {n2}")
else:
    print (f"The highest number is {n3}")
```
